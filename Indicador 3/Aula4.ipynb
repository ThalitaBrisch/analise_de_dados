```python
import pandas as pd
import openpyxl
import matplotlib as plt
import seaborn as sns
```

```python
caminho = r"H:\Python\Terceira Etapa\Aula 4\CD2022_Populacao_2010_Compatibilizada_20231222 (1).xlsx"

df = pd.read_excel(
    caminho,
    sheet_name="Municípios",
    skiprows=2
)

df.head()
```

```python
df = df.iloc[:, 1:]

df.columns = [
    "UF",
    "COD. UF",
    "COD. MUNIC",
    "NOME DO MUNICÍPIO",
    "População Município 2010",
    "População 2010 Compatibilizada",
    "População Censo 2022"
]

df.head()
```

```python
col_2010 = "População 2010 Compatibilizada"
col_2022 = "População Censo 2022"
```

```python
pop_estado = (
    df.groupby(["COD. UF", "UF"], as_index=False)
      .agg(
          populacao_2010=(col_2010, "sum"),
          populacao_2022=(col_2022, "sum"),
      )
)

pop_estado
```

```python
pop_estado["crescimento"] = (
    pop_estado["populacao_2022"] -
    pop_estado["populacao_2010"]
)
```

```python
pop_estado = pop_estado.sort_values(
    by="crescimento",
    ascending=False
)

pop_estado.head(10)
```

```python
top = pop_estado.head(10).copy()
```

```python
top = top.melt(
    id_vars="UF",
    value_vars=["populacao_2010", "populacao_2022"],
    var_name="ano",
    value_name="populacao"
)
```

```python
top["ano"] = top["ano"].replace({
    "populacao_2010": "2010",
    "populacao_2022": "2022"
})
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

sns.barplot(
    data=top,
    x="UF",
    y="populacao",
    hue="ano",
    ax=ax
)

ax.set_title(
    "10 UFs com maior crescimento absoluto: 2010 vs 2022"
)

ax.set_ylabel("Habitantes")
ax.set_xlabel("UF")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
