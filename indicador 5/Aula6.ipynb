```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
t2012 = pd.read_csv(
    "Tabela5-sem_emprego_2012.csv",
    sep=";"
)

t2026 = pd.read_csv(
    "Tabela5-sem_emprego_2026.csv",
    sep=";"
)
```

```python
t2012 = t2012.rename(columns={
    "Desocupados - mulheres (2012 T1)": "mulheres_2012"
})

t2026 = t2026.rename(columns={
    "Desocupados - mulheres (2026 T1)": "mulheres_2026"
})
```

```python
cuidados = pd.read_excel(r"H:\Python\Terceira Etapa\indicador 5\Tabela 1.1.1.xls")
```

```python
cuidados = cuidados[[
    "Sigla",
    "Código",
    "Estado",
    "Horas_cuidados"
]]
```

```python
comp = t2012.merge(
    t2026,
    on=["Sigla", "Código", "Estado"],
    how="inner"
)
```

```python
comp = comp.merge(
    cuidados,
    on=["Sigla", "Código", "Estado"],
    how="inner"
)
```

```python
ordem = comp.sort_values(
    "mulheres_2026"
)["Estado"]
```

```python
longo = comp.melt(
    id_vars=[
        "Sigla",
        "Código",
        "Estado",
        "Horas_cuidados"
    ],
    value_vars=[
        "mulheres_2012",
        "mulheres_2026"
    ],
    var_name="Ano",
    value_name="Participacao_mulheres"
)

longo["Ano"] = longo["Ano"].map({
    "mulheres_2012": "2012 T1",
    "mulheres_2026": "2026 T1"
})
```

```python
fig, ax = plt.subplots(figsize=(10,10))

sns.barplot(
    data=longo,
    y="Estado",
    x="Participacao_mulheres",
    hue="Ano",
    order=ordem,
    ax=ax
)

ax.axvline(
    50,
    color="red",
    linestyle="--",
    linewidth=1
)

ax.set_xlabel(
    "Participação das mulheres entre as pessoas desocupadas (%)"
)

ax.set_ylabel("")

ax.set_title(
    "Desocupação: participação feminina em 2012 T1 e 2026 T1"
)

ax.legend(title="Ano")

fig.tight_layout()

plt.show()
```
